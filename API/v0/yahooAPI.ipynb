```python
import yfinance as yf
```

```python
ticker = yf.Ticker("AAPL")
```

```python
info = ticker.info
financials = ticker.financials
balance_sheet = ticker.balance_sheet
cashflow = ticker.cashflow
print("financials:", financials)
print("balance sheet:", balance_sheet)
print("cash flow:", cashflow)
print("info object:", info)
print("Market Cap:", info.get("marketCap"))
print("PE Ratio:", info.get("trailingPE"))
```

```python
history = ticker.history(period="1d", interval="15m")
```

```python
print(history)
```

```python
news = ticker.news
```

```python
for n in news:
    print(n)
```
