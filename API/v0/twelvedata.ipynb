```python
from twelvedata import TDClient
import os
from dotenv import load_dotenv
import json

load_dotenv()

KEY = os.getenv("TWELVEDATA_KEY")
td = TDClient(apikey=KEY)
```

```python
ts = td.time_series(
    symbol="NVDA",
    interval="1h",
    outputsize=720,
)
```

```python
print(ts.as_csv())
```

```python
lp = td.price(symbol="AAPL")
```

```python
print(lp.as_json())
```
